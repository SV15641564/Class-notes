## 1. NumPy

### 1.1 Crear arreglos
Un `ndarray` guarda elementos **del mismo tipo** en un bloque contiguo de memoria; por eso es mucho más rápido que una lista de Python.

````python
a = np.array([10, 20, 30, 40, 50])
print(a, "| dtype:", a.dtype, "| forma:", a.shape, "| dimensiones:", a.ndim)

matriz = np.array([[1, 2, 3],
                   [4, 5, 6]])
print(matriz)
print("forma:", matriz.shape)   # (filas, columnas)

# Constructores frecuentes
print(np.zeros(3))               # ceros
print(np.ones((2, 2)))           # unos
print(np.arange(0, 10, 2))       # como range(), pero devuelve ndarray
print(np.linspace(0, 1, 5))      # 5 puntos equidistantes entre 0 y 1
````

### 1.2 Vectorización: operar sin bucles
**Algoritmo:** al escribir `a * 2`, NumPy ejecuta el bucle internamente en C, sobre memoria contigua. El resultado es el mismo que el de un `for`, pero mucho más rápido y más corto.

````python
import time

n = 1_000_000
lista = list(range(n))
arr = np.arange(n)

t0 = time.perf_counter()
doble_lista = [x * 2 for x in lista]       # bucle en Python
t1 = time.perf_counter()
doble_arr = arr * 2                         # vectorizado
t2 = time.perf_counter()

print(f"Lista : {t1 - t0:.4f} s")
print(f"NumPy : {t2 - t1:.4f} s")
print("¿Mismo resultado?", doble_lista == doble_arr.tolist())
````

### 1.3 Indexación, rebanado (*slicing*) y máscaras booleanas
- `a[i]` → un elemento; `a[i:j]` → del `i` al `j-1`; `a[::2]` → de 2 en 2.
- **Máscara booleana**: `a > 25` produce un arreglo de `True/False` del mismo tamaño; usarlo como índice **filtra**.

> **Predice:** ¿qué valores quedarán con `a > 25`? ¿Y con `(a > 15) & (a < 45)`?

````python
a = np.array([10, 20, 30, 40, 50])
print(a[0], a[-1], a[1:4], a[::2])

mascara = a > 25
print(mascara)          # [False False  True  True  True]
print(a[mascara])       # [30 40 50]

# Combinar condiciones: & (y), | (o), ~ (no). Cada condición va entre paréntesis.
print(a[(a > 15) & (a < 45)])

# Matrices: [fila, columna]
m = np.arange(12).reshape(3, 4)
print(m)
print(m[1, 2], m[:, 0], m[0, :])
````

### 1.4 *Broadcasting*: operar arreglos de formas distintas
**Regla del algoritmo:** NumPy compara las formas de derecha a izquierda. Dos dimensiones son compatibles si son **iguales** o una de ellas es **1** (la de tamaño 1 se "estira" virtualmente, sin copiar datos).

> **Predice:** ¿qué forma tendrá `m + fila`? ¿Y qué pasará al sumar un arreglo de forma `(2,)`?

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])         # forma (2, 3)

print(m + 10)                      # escalar -> se estira a (2, 3)

fila = np.array([100, 200, 300])   # forma (3,)  -> se estira a (2, 3)
print(m + fila)

col = np.array([[1000], [2000]])   # forma (2, 1) -> se estira a (2, 3)
print(m + col)

# Forma incompatible: (2,3) con (2,) -> error
try:
    m + np.array([1, 2])
except ValueError as e:
    print("Error esperado:", e)
````

### 1.5 Estadística y agregaciones
`sum`, `mean`, `std`, `min`, `max`, `median`... Con `axis` eliges en qué dirección colapsar: `axis=0` recorre **filas** (resultado por columna), `axis=1` recorre **columnas** (resultado por fila).

**Algoritmos a mano** (para entender qué hay detrás):
- Media: $\bar{x} = \frac{1}{n}\sum x_i$
- Desviación estándar: $\sigma = \sqrt{\frac{1}{n}\sum (x_i-\bar{x})^2}$ (NumPy usa `ddof=0`; pandas usa `ddof=1` por defecto).

````python
notas = np.array([[3.5, 4.0, 2.8],
                  [4.5, 4.2, 4.8],
                  [2.0, 3.1, 3.0],
                  [5.0, 4.9, 4.7]])   # 4 estudiantes x 3 exámenes

print("Media global      :", notas.mean())
print("Media por examen  :", notas.mean(axis=0))   # una por columna
print("Media por alumno  :", notas.mean(axis=1))   # una por fila
print("Máximo por examen :", notas.max(axis=0))
print("Alumno con mejor promedio (índice):", notas.mean(axis=1).argmax())

# Media y desviación "a mano" para verificar
x = notas[:, 0]
media = x.sum() / len(x)
desv = np.sqrt(((x - media) ** 2).sum() / len(x))
print(media, np.isclose(media, x.mean()), desv, np.isclose(desv, x.std()))
````

### 1.6 Algoritmo: normalización *z-score* y *min-max*
Escalar datos para compararlos es un paso típico antes de analizar o modelar.

1. **z-score:** $z = \dfrac{x-\bar{x}}{\sigma}$ → media 0 y desviación 1.
2. **min-max:** $x' = \dfrac{x-x_{min}}{x_{max}-x_{min}}$ → valores entre 0 y 1.

Gracias a *vectorización + broadcasting*, se aplican a toda la matriz en una línea.

````python
z = (notas - notas.mean(axis=0)) / notas.std(axis=0)     # por columna
mm = (notas - notas.min(axis=0)) / (notas.max(axis=0) - notas.min(axis=0))

print("z-score:\n", z.round(2))
print("media ≈", z.mean(axis=0).round(10), " std =", z.std(axis=0).round(10))
print("min-max:\n", mm.round(2))
````

### 3.1 Valores faltantes (`NaN`)
**Estrategias:** detectar (`isna`), **eliminar** (`dropna`) o **imputar** (`fillna`) con un valor razonable (media, mediana, constante).
**Cuidado:** la media se distorsiona con valores extremos; la **mediana** es más robusta.

````python
print(sucio.isna().sum())                       # faltantes por columna

sin_faltantes = sucio.dropna(subset=["nombre"])  # eliminar filas sin nombre
imput = sucio.copy()
imput["ingreso"] = imput["ingreso"].fillna(imput["ingreso"].median())
imput
````

### 3.2 Duplicados

### 3.4 Tipos de dato y fechas
- `pd.to_numeric(..., errors="coerce")` convierte y deja `NaN` donde no se pueda (así **no se rompe** el programa y el error queda visible).
- `pd.to_datetime` interpreta fechas; si hay formatos mezclados, conviene indicar `format=` y tratar los casos raros aparte.

````python
limpio["edad"] = pd.to_numeric(limpio["edad"], errors="coerce")       # "abc" -> NaN
limpio["ingreso_fecha"] = pd.to_datetime(limpio["ingreso_fecha"], format="%Y-%m-%d", errors="coerce")  # 05/02/2026 -> NaT
print(limpio.dtypes)
limpio
````

### 3.5 Pipeline completo de limpieza
Encadenar los pasos en una función hace el proceso **repetible** y fácil de revisar.

````python
def limpiar(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out = out.dropna(subset=["nombre"])
    out["nombre"] = out["nombre"].str.strip().str.title()
    out["ciudad"] = out["ciudad"].str.strip().str.title()
    out["edad"] = pd.to_numeric(out["edad"], errors="coerce")
    out["ingreso"] = out["ingreso"].fillna(out["ingreso"].median())
    out["ingreso_fecha"] = pd.to_datetime(out["ingreso_fecha"], format="%Y-%m-%d", errors="coerce")
    return out.drop_duplicates().reset_index(drop=True)

limpiar(sucio)
````

## 4. Agrupar y resumir: *split-apply-combine* (`groupby`)

> **Predice:** si agrupo las ventas por ciudad y sumo el `total`, ¿qué ciudad crees que vende más?

**Algoritmo en tres pasos:**
1. **Split (dividir):** separar las filas en grupos según los valores de una o varias columnas.
2. **Apply (aplicar):** calcular algo en cada grupo (suma, promedio, conteo...).
3. **Combine (combinar):** reunir los resultados en una nueva tabla.

```
 ciudad   total          Split            Apply (sum)        Combine
 Bogotá   100   ─►  Bogotá: [100, 50]  ─►     150     ─►   Bogotá  150
 Cali      70   ─►  Cali:   [70, 30]   ─►     100     ─►   Cali    100
 Bogotá    50
 Cali      30
```

````python
# Ejemplo mínimo hecho "a mano" con un diccionario, para ver el algoritmo
datos = [("Bogotá", 100), ("Cali", 70), ("Bogotá", 50), ("Cali", 30)]
acumulado = {}
for ciudad, total in datos:                  # split + apply
    acumulado[ciudad] = acumulado.get(ciudad, 0) + total
print(acumulado)                              # combine

# Lo mismo en pandas
mini = pd.DataFrame(datos, columns=["ciudad", "total"])
print(mini.groupby("ciudad")["total"].sum())
````

````python
# Con el dataset de ventas
print(ventas.groupby("ciudad")["total"].sum().sort_values(ascending=False))
````

````python
# Varias agregaciones a la vez con nombres claros (named aggregation)
resumen = ventas.groupby("producto").agg(
    ventas=("id_venta", "count"),
    unidades=("cantidad", "sum"),
    total=("total", "sum"),
    ticket_promedio=("total", "mean"),
).round(0).sort_values("total", ascending=False)
resumen
````

## 5. Combinar tablas: `merge` (uniones) y `concat`

### 5.1 `merge`: unir por una clave
Equivale al `JOIN` de SQL. **Algoritmo (*hash join*):** se construye un diccionario con la clave de la tabla derecha y, para cada fila de la izquierda, se busca su clave en ese diccionario (búsqueda ~ $O(1)$), en vez de comparar todas contra todas ($O(n \cdot m)$).

| `how` | Qué conserva |
|---|---|
| `inner` | Solo claves presentes en **ambas** |
| `left` | Todas las de la izquierda |
| `right` | Todas las de la derecha |
| `outer` | Todas las de ambas |

````python
clientes = pd.DataFrame({
    "id_cliente": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Marta", "Pedro"],
})
pedidos = pd.DataFrame({
    "id_pedido": [101, 102, 103, 104],
    "id_cliente": [1, 1, 3, 9],      # el 9 no existe en clientes
    "valor": [50, 70, 20, 99],
})

print("inner:\n", clientes.merge(pedidos, on="id_cliente", how="inner"))
print("left:\n", clientes.merge(pedidos, on="id_cliente", how="left"))
print("outer:\n", clientes.merge(pedidos, on="id_cliente", how="outer", indicator=True))
````

**Hash join a mano**, para ver que no hay magia:

````python
indice = {}                                     # fase 1: construir el diccionario con la tabla derecha
for fila in pedidos.to_dict("records"):
    indice.setdefault(fila["id_cliente"], []).append(fila)

resultado = []                                   # fase 2: recorrer la izquierda y buscar
for c in clientes.to_dict("records"):
    for p in indice.get(c["id_cliente"], []):
        resultado.append({**c, **p})
pd.DataFrame(resultado)                          # = inner join
````

### 5.2 `concat`: apilar tablas
`axis=0` apila filas (mismas columnas); `axis=1` pega columnas lado a lado (alineando por índice).

````python
enero = pd.DataFrame({"producto": ["Lápiz", "Cuaderno"], "unidades": [10, 5]})
febrero = pd.DataFrame({"producto": ["Lápiz", "Mochila"], "unidades": [8, 2]})
pd.concat([enero.assign(mes="ene"), febrero.assign(mes="feb")], ignore_index=True)
````

## 6. Mini-proyecto integrador
**Pregunta de negocio:** ¿qué ciudad y producto generan más ingresos y cómo evolucionan por mes?

Flujo: *cargar → limpiar → transformar → agrupar → concluir*.

````python
# 1) "Cargar": en un caso real sería pd.read_csv("archivo.csv")
df = ventas.copy()

# 2) Limpiar y validar
assert df["id_venta"].is_unique, "id_venta duplicado"
assert df[["cantidad", "precio_unitario"]].gt(0).all().all(), "valores no positivos"

# 3) Transformar
df["ingreso"] = df["cantidad"] * df["precio_unitario"]

# 4) Agrupar
top = (df.groupby(["ciudad", "producto"])["ingreso"].sum()
         .sort_values(ascending=False).head(5))
print("Top 5 combinaciones ciudad-producto:\n", top)

por_mes = df.pivot_table(values="ingreso", index="mes", columns="region", aggfunc="sum", fill_value=0)
por_mes
````

````python
# 5) Visualizar (pandas usa matplotlib por debajo)
por_mes.plot(kind="bar", figsize=(7, 3.5), title="Ingresos por mes y región (datos sintéticos)")
````

**Conclusión del ejemplo:** el resultado depende de datos aleatorios, pero el *método* es el que importa: validar → transformar → agrupar → comunicar.

## 7. Buenas prácticas y errores comunes
- **Evita bucles `for` sobre filas**: casi siempre hay una operación vectorizada (`df["a"] * df["b"]`, `np.where`, `groupby`).
- **Copias vs. vistas:** usa `.copy()` antes de modificar un subconjunto, para no afectar el original ni recibir advertencias.
- **Condiciones combinadas:** usa `&`, `|`, `~` con paréntesis; no uses `and`/`or` entre Series.
- **Nunca confíes ciegamente en los datos:** revisa `info()`, `describe()`, `isna().sum()` y duplicados antes de analizar.
- **`loc` es por etiqueta (incluye el final), `iloc` es por posición (excluye el final).**
- **Reproducibilidad:** fija semillas y deja el código en funciones.
- **Datos sensibles:** trabaja con datos sintéticos o anonimizados cuando se trate de personas.

### NumPy
| Concepto | Idea clave |
|---|---|
| `ndarray` | Arreglo homogéneo y contiguo en memoria → rápido |

| Concepto | Idea clave |
|---|---|
| `Series` / `DataFrame` | Arreglos NumPy con etiquetas; tabla de columnas |
| Exploración | `head`, `info`, `describe`, `value_counts` |

| `pivot_table` | `groupby` con dos claves en formato matriz |

### Algoritmos explicados
1. **Vectorización** – el bucle ocurre en C sobre memoria contigua.
2. **Broadcasting** – comparación de formas de derecha a izquierda.
3. **Media y desviación estándar** – fórmulas implementadas a mano y verificadas.
4. **Normalización** (z-score y min-max).
5. **Máscara booleana** – filtrar con un arreglo de `True/False`.
6. **Ordenamiento** – `sort_values` de una o varias claves.
7. **Split-apply-combine** – implementado con un diccionario y con `groupby`.
8. **Hash join** – implementado a mano y con `merge`.
